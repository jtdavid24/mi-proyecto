```python
import pandas as pd
import plotly.express as px

# Cargar el dataset
df = pd.read_csv('../vehicles_us.csv')


# Vista general
df.head()
```

## Grafico de Distribución de precios

```python
px.histogram(df, x='price', nbins=50, title='Distribución de precios de vehículos')
```

## Número de vehículos por tipo

```python
vc = df['type'].value_counts()

px.bar(vc, x=vc.index, y=vc.values,
       title='Cantidad de vehículos por tipo',
       labels={'x': 'Tipo de vehículo', 'y': 'Cantidad'})
```

## Año del modelo vs precio

```python
# Asegurar que los datos sean numéricos y sin valores nulos
df_clean = df[['model_year', 'price', 'type']].dropna()
df_clean['model_year'] = df_clean['model_year'].astype(int)
df_clean['price'] = df_clean['price'].astype(float)

# Crear gráfica
px.scatter(df_clean, x='model_year', y='price', color='type',
           title='Precio vs Año del Modelo',
           labels={'model_year': 'Año del Modelo', 'price': 'Precio'},
           opacity=0.5)
```
